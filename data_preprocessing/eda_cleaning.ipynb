```python
df=pd.read_csv('../data/loans.csv')
```

```python
df.loc[df['fico'] > 850, 'fico'] = df.loc[df['fico'] > 850, 'fico'] - 1000
```

```python
df.loc[df['int.rate'] > 1, 'int.rate'] = df.loc[df['int.rate'] > 1, 'int.rate'] / 100
```

```python
df['fico'].describe()
```

```python
df['int.rate'].describe()
```

```python
purpose_dummies = pd.get_dummies(df['purpose'], prefix='purpose', drop_first=True).astype(int)
df = pd.concat([df, purpose_dummies], axis=1)
df = df.drop(columns=['purpose'])
```

```python
df['delinq.2yrs'] = df['delinq.2yrs'].replace('yes', 1)
df['delinq.2yrs'] = pd.to_numeric(df['delinq.2yrs'])
print(df['delinq.2yrs'].dtype)
```

```python
df[pd.to_numeric(df['pub.rec'], errors='coerce').isna()]
```

```python
df['pub.rec'] = df['pub.rec'].replace('no', 0)
df['pub.rec'] = pd.to_numeric(df['pub.rec'])
print(df['pub.rec'].dtype)
```
